```python
from datasets import load_dataset, load_from_disk

def write_d(dataset, path):
    dataset.to_parquet(path)
    print(f"[INFO] finish write {path}", flush=True)
```

```python
path = "/home/ma-user/work/dev/_huggingface/_datasets/_data2/processed_dataset_new/luffy_train_unfilter_sys_wo_format/train_with_uuid.parquet"
dataset = load_dataset("parquet", data_files=path)['train']
```

```python
dataset[0].keys()

dataset[0]['reward_model']
```

```python
def map_func(line):
    uuid = line['uuid']
    return {
        'uid': uuid
    }

dataset = dataset.map(map_func)
```

```python
# 假设 dataset 已经是你经过上一步处理后的 Dataset 对象

# 1. 选取前 256 条数据 (如果数据总量不足 256，则选取全部)
n_samples = 256
top_dataset = dataset.select(range(min(len(dataset), n_samples)))

# 2. 定义处理函数：将 'uid' 赋值给 'data_source'
def assign_source(example):
    example['data_source'] = example['uid']
    return example

# 3. 应用映射
final_dataset = top_dataset.map(assign_source)

# 打印第一条数据检查结果
print("数据量:", len(final_dataset))
print("示例数据:", final_dataset[0])
```

```python
# 设定重复次数
n_repeats = 128

# 1. 生成重复的索引列表: [0, 0... (128次), 1, 1... (128次), ...]
# 这样可以保证数据是按单条扩充的 (A, A, A, B, B, B...)
repeated_indices = [i for i in range(len(final_dataset)) for _ in range(n_repeats)]

# 2. 使用 select 根据索引重组数据集
expanded_dataset = final_dataset.select(repeated_indices)

# 打印结果验证
print(f"原始数量: {len(final_dataset)}")
print(f"扩充后数量: {len(expanded_dataset)} (应为 {len(final_dataset)} * 128 = {len(final_dataset)*128})")

# 检查前3条数据是否相同
print("前3条数据的uid:", expanded_dataset[:3]['uid'])

write_d(expanded_dataset, "/home/ma-user/work/dev/_huggingface/_datasets/_psrnsr/val256_mean128.parquet")
```
