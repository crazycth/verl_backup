## Sample Val Set

```python
from datasets import load_dataset
from pprint import pprint
from collections import defaultdict
from tqdm import tqdm
```

```python
val_files="/home/ma-user/work/dev/_huggingface/_datasets/_data2/processed_dataset_new/luffy_test_sys_wo_format/test.parquet"

val = load_dataset("parquet", data_files=val_files)['train']
```

```python
from collections import defaultdict

counts = defaultdict(int)

for line in val:
    s = line['data_source']
    counts[s] += 1

for k,v in counts.items():
    print(k, v)
```

```python
# 只保留 data_source ∈ {"aime", "aime25"}
keep_sources = {"aime", "aime25"}

val = val.filter(lambda ex: ex.get("data_source") in keep_sources)

print("filtered val size:", len(val))
print("filtered data_source counts:")
from collections import Counter
print(Counter(val["data_source"]))
```
